```python
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
import numpy as np

```

```python
train_data_dir = 'train'
validation_data_dir = 'val'
train_dataset = tf.keras.preprocessing.image_dataset_from_directory(train_data_dir,image_size=(128,128),batch_size=32,shuffle=True,seed=42)
validation_dataset = tf.keras.preprocessing.image_dataset_from_directory(validation_data_dir,image_size=(128,128),batch_size=32,shuffle=True,seed=42)
```

```python
plt.figure(figsize=(10,10))

class_name = train_dataset.class_names

for images,labels in train_dataset.take(1):
  for i in range(9):
    ax = plt.subplot(3,3,i+1)
    plt.imshow(images[i].numpy().astype('uint8'))
    plt.title(class_name[labels[i]])
    plt.axis("off")
    plt.show()
```
